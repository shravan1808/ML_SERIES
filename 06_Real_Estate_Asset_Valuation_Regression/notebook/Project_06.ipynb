```python
final_df = pd.merge(df_listings, df_neighborhood, on='Neighborhood_ID')
final_df = pd.merge(final_df, df_macro, on='Property_ID')
```

```python
print(final_df.head().to_string())
```

```python
features = ['Area_SqFt','Bedrooms','Building_Age_Yrs','Distance_City_KM','School_Rating','Crime_Index','Interest_Rate_Pct','Property_Tax_Rate']
target = 'Price_USD_M'
```

```python
row = final_df.shape[0]
col = final_df.shape[1]
print(f"{'Total Records':<25} : {row}")
print(f"{'Total Combined Columns':<25} : {col}")
print(f"{'Target Variable':<25} : {target}")
print(f"{'Target Data Type':<25} : {final_df[target].dtype}")
print(f"{'Supervised Paradigm':<25} : Regression (Continuous Target)")
```

```python
print(f"Features (X) columns: ")
for col in features:
  print(f"- {col}")
print(f"\nTarget (y): ")
print(f"- {target}")
```

```python
corr = final_df[features + [target] ].corr()[target]
print(corr[features].sort_values(ascending=False).to_string(float_format='%.3f'))
print(f"\nStrongest Positive Feature : {corr[features].idxmax()} ({round(corr[features].max(),3)})")
print(f"Strongest Negative Feature : {corr[features].idxmin()} ({round(corr[features].min(),3)})")
```

```python
high_value = final_df[final_df['Price_USD_M']>1.5]
moderate_value = final_df[final_df['Price_USD_M']<=1.5]
high_val_shape = high_value.shape[0]
moderate_val_shape = moderate_value.shape[0]

high_mean_area = round(high_value['Area_SqFt'].mean(),2)
moderate_mean_area = round(moderate_value['Area_SqFt'].mean(),2)

high_mean_distance = round(high_value['Distance_City_KM'].mean(),2)
moderate_mean_distance = round(moderate_value['Distance_City_KM'].mean(),2)

high_mean_school = round(high_value['School_Rating'].mean(),2)
moderate_mean_school = round(moderate_value['School_Rating'].mean(),2)


print("High Value Properties (> $1.5M) :")
print(f"-{'Total Properties':<20} : {high_val_shape}")
print(f"-{'Mean Area':<20} : {high_mean_area} SqFt")
print(f"-{'Mean Distance':<20} : {high_mean_distance} KM")
print(f"-{'Mean School Rating':<20} : {high_mean_school}")

print("\nModerate Value Properties (<= $1.5M) :")
print(f"-{'Total Properties':<20} : {moderate_val_shape}")
print(f"-{'Mean Area':<20} : {moderate_mean_area} SqFt")
print(f"-{'Mean Distance':<20} : {moderate_mean_distance} KM")
print(f"-{'Mean School Rating':<20} : {moderate_mean_school}")
```

```python
print('''REGRESSION ALGORITHM MAP:
1. Linear Regression    : Baseline linear relationship mapping; fast and interpretable.
2. Ridge / Lasso        : Handles multicollinearity between Area_SqFt and Bedrooms.
3. Decision Tree Regressor : Captures non-linear local neighborhood threshold effects.
''')
```
