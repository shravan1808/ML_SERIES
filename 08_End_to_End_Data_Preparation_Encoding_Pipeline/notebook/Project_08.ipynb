```python
final_df = (
    df_profile
    .merge(df_tx,on='Customer_ID')
    .merge(df_engagement,on='Customer_ID')
)
```

```python
initial_records = final_df.shape[0]
initial_columns = final_df.shape[1]
```

```python
print(final_df.head().to_string())
```

```python
missing_values_before = final_df.isna().sum()
missing_values = missing_values_before[missing_values_before!=0]
print("Missing Values Before Imputation:")
for missing_col,row in missing_values.items():
  if row > 1:
    print(f"- {missing_col:<20} : {row} missing values")
  else:
    print(f"- {missing_col:<20} : {row} missing value")

med_values = final_df.groupby('Account_Tier')[['Age','Total_Spend_USD']].median()
print(f"\nImputed Metrics by Account_Tier (Age / Spend Medians):")
for tier,row in med_values.iterrows():
  print(f"- {tier:<8} : Median Age = {row['Age']:.1f} | Median Spend = ${row['Total_Spend_USD']:.2f}")

final_df['Age'] = final_df['Age'].fillna(final_df.groupby('Account_Tier')['Age'].transform('median'))
final_df['Total_Spend_USD'] = final_df['Total_Spend_USD'].fillna(final_df.groupby('Account_Tier')['Total_Spend_USD'].transform('median'))

missing_values_after = final_df.isna().sum().sum()
print(f"\nMissing Values After Imputation: {missing_values_after}")
```

```python
tier_mapping = {
    'Bronze':1,
    'Silver':2,
    'Gold':3,
    'Platinum':4
}

final_df['Account_Tier_Encoded'] = final_df['Account_Tier'].map(tier_mapping)

region_encoded = pd.get_dummies(final_df['Region'],
                                prefix='Region',
                                drop_first=True,
                                dtype=int)

final_df= pd.concat([final_df,region_encoded],axis=1)

print("Encoded Columns:")
print(
    f"- Account_Tier_Encoded : "
    f"Discrete Values {sorted(final_df['Account_Tier_Encoded'].unique().tolist())}"
)

region_columns = [
    col for col in final_df.columns
    if col.startswith('Region_')
]

print(
    f"- One-Hot Region Columns: "
    f"{', '.join(region_columns)} "
    f"(Region_East dropped as baseline)"
)
```

```python
Q1 = final_df['Total_Spend_USD'].quantile(0.25)
Q3 = final_df['Total_Spend_USD'].quantile(0.75)
IQR = Q3-Q1
UB = Q3 + (1.5*IQR)
outlier_count = (final_df['Total_Spend_USD']>UB).sum()
final_df['Total_Spend_USD'] = final_df['Total_Spend_USD'].clip(upper=UB)
print('IQR Thresholds for Total_Spend_USD:')
print(f"- {'Q1 (25th Percentile)':<20} : ${Q1:.2f}")
print(f"- {'Q3 (75th Percentile)':<20} : ${Q3:.2f}")
print(f"- {'IQR':<20} : ${IQR:.2f}")
print(f"- {'Upper Bound (Q3+1.5*IQR)':<20} : ${UB:.2f}")
print(f"\n{'Outliers Detected':<22} : {outlier_count} records capped to {UB:.2f}")
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
continuous_features =  [
    'Age',
    'Total_Spend_USD',
    'Purchase_Frequency',
    'App_Sessions_Per_Month'
]

scaled_features = scaler.fit_transform(final_df[continuous_features])

scaled_df = pd.DataFrame(
              scaled_features,
              columns= ['Scaled_Age',
                        'Scaled_Total_Spend_USD',
                        'Scaled_Purchase_Frequency',
                        'Scaled_App_Sessions'],
              index = final_df.index
              )

final_df = pd.concat([final_df,scaled_df] , axis =1)

print("Standard Scaler Metrics:")

for col in scaled_df.columns:
  print(f"- {col:<30} : Mean = {scaled_df[col].mean():.2f}, Std = {scaled_df[col].std(ddof=0):.2f} ")

```
