```python
final_df = pd.merge(df_shipments,df_fleet,on='Vehicle_ID')
final_df = pd.merge(final_df,df_weather,on='Region_ID')
```

```python
row_count= final_df.shape[0]
col_count = final_df.shape[1]
```

```python
final_df.info()
```

```python
features = ['Distance_KM','Cargo_Weight_Tons','Actual_Delay_Hours','Vehicle_Age_Years','Maintenance_Score','Driver_Experience_Yrs','Weather_Condition','Traffic_Congestion_Index']
target = ['Is_Delayed']
```

```python
tar_imbalance = final_df['Is_Delayed'].value_counts()
```

```python
distance_km_mean= final_df.groupby('Is_Delayed')['Distance_KM'].mean()
distance_km_mean
```

```python
vehicle_main_score = final_df.groupby('Is_Delayed')['Maintenance_Score'].mean()
vehicle_main_score
```

```python
final_df['Prediction'] = np.where((final_df['Distance_KM'] > 400) & (final_df['Maintenance_Score'] < 60) | (final_df['Weather_Condition'].isin(['Storm', 'Heavy Rain'])), 'Yes', 'No')
```

```python
total_pred = len(final_df['Prediction'])
correct_pred = (final_df['Prediction'] == final_df['Is_Delayed']).sum()
fn=((final_df['Is_Delayed'] == 'Yes') & (final_df['Prediction'] == 'No')).sum()
fp=((final_df['Is_Delayed'] == 'No') & (final_df['Prediction'] == 'Yes')).sum()
tn=((final_df['Is_Delayed'] == 'No') & (final_df['Prediction'] == 'No')).sum()
tp=((final_df['Is_Delayed'] == 'Yes') & (final_df['Prediction'] == 'Yes')).sum()
rule_accuracy =((tp+tn)/total_pred)*100
```

```python
print(f"{'='*10} ENTERPRISE LOGISTICS DELAY ANALYTICS {'='*10}")
print(f"\nIntegrated Dataset Records  : {row_count}")
print(f"Total Combined Features : {col_count}")
print(f"Target Class Distribution  : No = {tar_imbalance.iloc[0]} ({(tar_imbalance.iloc[0]/row_count)*100}%), Yes = {tar_imbalance.iloc[1]} ({(tar_imbalance.iloc[1]/row_count)*100}%)")
print(f"\nGroup Metrics (Delayed vs On-Time):")
print(f"{'-Mean Distance':<25} : Delayed = {distance_km_mean.iloc[0]:.2f} KM | On-Time = {distance_km_mean.iloc[1]:.2f} KM")
print(f"{'-Mean Vehicle Maintenance':<25} : Delayed = {vehicle_main_score.iloc[0]:.2f} Score | On-Time = {vehicle_main_score.iloc[1]:.2f} Score")
print("\nRule Engine Evaluation:")
print(f"{'-Rule Accuracy':<25} : {rule_accuracy:.2f}%")
print(f"{'-False Positives':<25} : {fp}")
print(f"{'-False Negatives':<25} : {fn}")
print("\nConclusion:")
print('''Multi-table integration reveals that severe weather and
low maintenance score interact non-linearly with distance. An ML
algorithm dynamically weights these interactions without hand-coded threshold rules.''')
```
