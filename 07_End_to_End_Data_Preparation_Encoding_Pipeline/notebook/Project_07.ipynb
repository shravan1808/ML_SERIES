```python
final_df = df_activity.merge(df_sub, on='User_ID').merge(df_device, on='Device_ID')
```

```python
print(final_df.head().to_string())
```

```python
df_shape = final_df.shape
print(f"{'Total Combined Records':<25} : {df_shape[0]}")
print(f"{'Total Combined Columns':<25} : {df_shape[1]}")
print(f"{'Target Vector (y)':<25} : ABSENT (Unsupervised Setup)")
print(f"{'Paradigm':<25} : Unsupervised Learning / Clustering")
```

```python
print("Numeric Features (X) Range Analysis:")

min_daily_lis_mins = final_df['Daily_Listening_Mins'].min()
max_daily_lis_mins = final_df['Daily_Listening_Mins'].max()
min_playlists_Created = final_df['Playlists_Created'].min()
max_playlists_created = final_df['Playlists_Created'].max()
min_skip_ratio = final_df['Skip_Ratio'].min()
max_skip_ratio = final_df['Skip_Ratio'].max()
min_podcast_lis_pct = final_df['Podcast_Listening_Pct'].min()
max_podcast_lis_pct = final_df['Podcast_Listening_Pct'].max()
min_account_age_months = final_df['Account_Age_Months'].min()
max_account_age_months = final_df['Account_Age_Months'].max()

print(f"{'Daily_Listening_Mins':<25} : Min = {min_daily_lis_mins}, Max = {max_daily_lis_mins} (Span: {max_daily_lis_mins-min_daily_lis_mins})")
print(f"{'Playlists_Created':<25} : Min = {min_playlists_Created}, Max = {max_playlists_created} (Span: {max_playlists_created-min_playlists_Created})")
print(f"{'Skip_Ratio':<25} : Min = {min_skip_ratio}, Max = {max_skip_ratio} (Span: {max_skip_ratio-min_skip_ratio})")
print(f"{'Podcast_Listening_Pct':<25} : Min = {min_podcast_lis_pct}, Max = {max_podcast_lis_pct} (Span: {max_podcast_lis_pct-min_podcast_lis_pct})")
print(f"{'Account_Age_Months':<25} : Min = {min_account_age_months}, Max = {max_account_age_months} (Span: {max_account_age_months-min_account_age_months})")

print("""\nScale Analysis: Daily_Listening_Mins dominates distance metrics due to magnitude.
Feature scaling (StandardScaler/MinMaxScaler) is mandatory.""")
```

```python
final_df['Listeners'] = np.where((final_df['Daily_Listening_Mins']>100) & (final_df['Monthly_Fee_USD']>0),'Power','Casual')
power_list = (final_df['Listeners']=='Power').sum()
casual_list = (final_df['Listeners']=='Casual').sum()
print("Heuristic Segment Counts:")

print(f"{'Cluster 0 (Power Listeners)':<28} : {power_list} ({power_list/df_shape[0]*100}%) ")
print(f"{'Cluster 1 (Casual Listeners)':<28} : {casual_list} ({casual_list/df_shape[0]*100}%)")
```

```python
features =['Daily_Listening_Mins','Playlists_Created','Skip_Ratio','Podcast_Listening_Pct','Account_Age_Months']
cluster_profile_mean = final_df.groupby('Listeners')[features].mean()
print("Mean Profile by Cluster:")
print(cluster_profile_mean.to_string())
```
