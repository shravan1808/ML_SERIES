```python
data = """Customer_ID,Account_Age_Days,Monthly_Spend,Support_Calls,Inactivity_Days,Churn
C101,120,45.5,4,35,Yes
C102,450,120.0,0,5,No
C103,90,25.0,5,40,Yes
C104,300,85.0,1,12,No
C105,210,60.0,2,20,No
C106,60,15.0,6,45,Yes
C107,500,200.0,1,8,No"""
```

```python
with open('02_Ecommerce_Customer_Churn_Prediction/data/customer_data.csv','w') as file:
  file.write(data)
```

```python
cust_df=pd.read_csv('02_Ecommerce_Customer_Churn_Prediction/data/customer_data.csv')
```

```python
print(cust_df.to_string())
```

```python
print(f"{'Number of Rows':<18}: {cust_df.shape[0]}")
print(f"{'Number of Columns':<18}: {cust_df.shape[1]}")
print('\nColumns:')
for col in cust_df.columns:
  print(col)
```

```python
features = ['Account_Age_Days','Monthly_Spend','Support_Calls','Inactivity_Days']
target = ['Churn']
print("Features:")
for feature in features:
  print(feature)
print("\nTarget:")
print(target[0])
```

```python
print("Churn Status Counts:")
churn_count=cust_df['Churn'].value_counts()
print(f"No : {churn_count.iloc[0]}")
print(f"Yes : {churn_count.iloc[1]}")
```

```python
churn_cust=cust_df[cust_df['Churn']=='Yes']
ret_cust=cust_df[cust_df['Churn']=='No']
churn_cust_avg=churn_cust['Inactivity_Days'].mean()
ret_cust_avg=ret_cust['Inactivity_Days'].mean()
print("Average Inactivity Days:")
print(f"{'Churned Customers':<20}: {churn_cust_avg} days")
print(f"{'Returned Customers':<20}: {ret_cust_avg} days")
```

```python
cust_df['Prediction']=np.where(((cust_df['Inactivity_Days']>30) & (cust_df['Monthly_Spend']<50)),'Yes','No')
print(cust_df[['Customer_ID','Churn','Prediction']])
```

```python
correct_pred = np.where(cust_df['Churn']==cust_df['Prediction'])
print(f"{'Correct Predictions':<20}: {len(correct_pred[0])}")
print(f"{'Total Predictions':<20}: {len(cust_df['Prediction'])}")
print(f"{'Rule-Based Accuracy':<20}: {(len(correct_pred[0])/len(cust_df['Prediction'])*100)}%")
```

```python
cust_A = {
    'Inactivity_Days' : 32,
    'Monthly_Spend' : 150.0,
    'Support_Calls' : 6
}
cust_B = {
    'Inactivity_Days' : 28,
    'Monthly_Spend' : 30.0,
    'Support_Calls' : 8
}
```

```python
cust_A['Pred'] = np.where(((cust_A['Inactivity_Days']>30) & (cust_A['Monthly_Spend']<50)),'Yes','No')
cust_B['Pred'] = np.where(((cust_B['Inactivity_Days']>30) & (cust_B['Monthly_Spend']<50)),'Yes','No')

print(f"{'Customer A Prediction':<20} : {cust_A['Pred']}")
print(f"{'Customer B Prediction':<20} : {cust_B['Pred']}")
```

```python
print(f"{'='*10} CUSTOMER CHURN ANALYSIS {'='*10}")
print()
print(f"{'Total Customers':<20} : {len(cust_df)}")
print(f"{'Churned Customers':<20} : {len(churn_cust)}")
print(f"{'Retained Customers':<20} : {len(ret_cust)}")
print("\nAverage Inactivity")
print(f"{'Churned Customers':<20} : {churn_cust_avg} days")
print(f"{'Retained Customers':<20} : {ret_cust_avg} days")
print(f"\n{'Rule-Based Accuracy':<20} : {(len(correct_pred[0])/len(cust_df['Prediction'])*100)}%")
print(f"\n{'New Customer A Pred':<20} : {cust_A['Pred']}")
print(f"{'New Customer B Pred':<20} : {cust_B['Pred']}")
print()
print('''Conclusion:
Hardcoded rules fail when missing complex interactions (like high support calls).
ML models dynamically balance all input features simultaneously.''')
```
