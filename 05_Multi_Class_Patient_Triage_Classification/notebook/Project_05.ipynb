```python
import pandas as pd
import numpy as np
```

```python
pat_df=pd.read_csv('patients.csv')
```

```python
pat_df.head()
```

```python
pat_df.info()
```

```python
df_size = pat_df.shape
```

```python
pat_df.describe()
```

```python
X = ['Heart_Rate_BPM','Systolic_BP','Oxygen_Sat_Pct','Age']
y = 'Triage_Category'
```

```python
tar_count = pat_df[y].value_counts()
tar_pct = tar_count/df_size[0]*100
```

```python
triage_level_mean = pat_df.groupby(y)[X].mean().reset_index()
triage_level_mean
```

```python
print(f"{'='*10} EMERGENCY PATIENT TRIAGE ANALYSIS {'='*10}")
print(f"\n{'Dataset Shape':<27} : {df_size[0]} Rows, {df_size[1]} Columns")
print(f"{'Supervised Paradigm':<27} : Multi Class Classification (Discrete Categorical Target)")
print(f"\n{'Features (X)':<27} : {X[0]},{X[1]},{X[2]},{X[3]}")
print(f"{'Target (y)':<27} : {y}")
print(f"\n{'Class Distribution':<27} :")
for category in ['General', 'Urgent', 'ICU']:
    count = tar_count[category]
    percentage = tar_pct[category]
    print(f"-{category:<26} : {count} ({percentage:.2f}%)")
print(f"\n{'Triage Level Group Averages':<25} :")
print(triage_level_mean.to_string(index= False))
print("\nAlgorithm Guidance: Decision Trees or Multi-Class Logistic Regression are recommended for interpretable clinical rule boundaries.")
```
