```
#Task 1
final_df = df_tx.merge(df_acc,on='Account_ID')
final_df = final_df.merge(df_trm,on='Terminal_ID')
final_df = final_df.merge(df_blk,on = 'Tx_ID')
```

```
final_df.head()
```

```
row_count = final_df.shape[0]
col_count = final_df.shape[1]
final_df.info()
print(final_df.isna().sum())
```

```
features = ['Tx_Amount','Tx_Hour','Avg_Monthly_Tx_Vol','Account_Age_Months','Terminal_Risk_Score','Is_Foreign_Location','High_Risk_IP_Flag']
target = 'Is_Fraud'
```

```
target_dist = final_df[target].value_counts()
target_dist
```

```
med_tx_amt = final_df.groupby(target)['Tx_Amount'].median()
med_trm_risk_score = final_df.groupby(target)['Terminal_Risk_Score'].median()
```

```
final_df['Prediction'] = np.where(((final_df['Tx_Amount']>1000) & (final_df['Tx_Hour'].isin([1,2,3,4]))) | (final_df['High_Risk_IP_Flag'] == 1),1,0)
```

```
final_df[['Is_Fraud','Prediction']].head(10)
```

```
fp = ((final_df['Is_Fraud']==0) & (final_df['Prediction']==1)).sum()
fn = ((final_df['Is_Fraud']==1) & (final_df['Prediction']==0)).sum()
tn = ((final_df['Is_Fraud']==0) & (final_df['Prediction']==0)).sum()
tp = ((final_df['Is_Fraud']==1) & (final_df['Prediction']==1)).sum()
correct_pred = tp+tn
total_predictions = tp+tm+fp+fn
accuracy = ((tp+tn)/total_predictions)*100
```
