SMART EMPLOYEE ATTENDANCE PREDICTION

```python
import pandas as pd
import numpy as np
```

```python
print(pd.__version__)
print(np.__version__)
```

```python
with open('employee_det','w') as file:
    file.write('''Employee_ID,Distance_KM,Travel_Time_Min,Previous_Late_Count,Weather,Transport,Late
                  E101,5,20,1,Clear,Bus,No
                  E102,18,65,5,Rain,Bus,Yes
                  E103,7,30,0,Clear,Car,No
                  E104,22,75,7,Rain,Bus,Yes
                  E105,10,40,2,Cloudy,Bike,No
              ''')
```

```python
df=pd.read_csv('employee_det')
```

```python
print(df.to_string())
```

```python
print(f'{'Number of Rows':20}: {df.shape[0]}')
print(f"{'Number of Columns':20}: {len(df.columns)}")
print("\nColumns:")
for col in df.columns:
    print(col)
```

```python
features = ['Distance_KM', 'Travel_Time_Min', 'Previous_Late_Count','Weather','Transport']
target = ['Late']
print("Features:")
for feature in features:
    print(feature)
print("\nTarget:")
print(target[0])
```

```python
late_emp_count = df[['Late']].value_counts()
print(late_emp_count)
print()
print(f'{'Late Employees':20}:{late_emp_count.iloc[0]}')
print(f'{"Not Late Employees":20}:{late_emp_count.iloc[1]}')
```

```python
late_emp = df[df['Late']=='Yes']
not_late_emp = df[df['Late']=='No']
late_emp_avg=late_emp['Travel_Time_Min'].mean()
not_late_emp_avg=not_late_emp['Travel_Time_Min'].mean()
print("Average Travel Time")
print()
print("Late Employees:")
print(f"{late_emp_avg:.1f} minutes")
print()
print("Not Late Employees:")
print(f"{not_late_emp_avg:.1f} minutes")
```

```python
df['Prediction']=np.where((df['Distance_KM']>15) & (df['Travel_Time_Min']> 60),'Yes','No')
print(df[['Employee_ID','Late','Prediction']].to_string(index=False))
```

```python
total_predictions = df['Prediction'].count()
correct_predictions = (df['Prediction'] == df['Late']).sum()
rule_accuracy = (correct_predictions / total_predictions)*100
print(f"{'Correct Predictions':20}: {correct_predictions}")
print(f"{'Total Predictions':20}: {total_predictions}")
print()
print(f"{'Rule-Based Accuracy':20}:{int(rule_accuracy)}%")
```

```python
new_emp1 = {'Distance_KM' : 20,
'Travel_Time_Min' : 70,
'Previous_Late_Count' : 3,
'Weather' : 'Rain',
'Transport' : 'Bus'}

new_emp1['Prediction']=np.where(new_emp1['Distance_KM']>15 and new_emp1['Travel_Time_Min']> 60,'Yes','No')
print(f"{'Employee Distance':20}: {new_emp1['Distance_KM']} KM")
print(f"{'Travel Time':20}: {new_emp1['Travel_Time_Min']} Minutes\n")
print(f"{'Prediction':20}: {new_emp1['Prediction']}")
```
